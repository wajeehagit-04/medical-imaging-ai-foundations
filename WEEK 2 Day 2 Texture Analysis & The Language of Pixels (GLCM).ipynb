```python
# 1. Install dependencies first
!pip install SimpleITK numpy pandas

# 2. Install the latest development version of pyradiomics from GitHub
# This version contains the fixes for Python 3.12
!pip install git+https://github.com/AIM-Harvard/pyradiomics.git@master
import SimpleITK as sitk
import pandas as pd
from radiomics import getTestCase, glcm

# 1. Load the sample brain data
# (If you use your own data, replace these with your file paths)
imagePath, maskPath = getTestCase('brain1')
image = sitk.ReadImage(imagePath)
mask = sitk.ReadImage(maskPath)

# 2. Initialize the GLCM (Gray-Level Co-occurrence Matrix) Calculator
# We set binWidth=25 to group similar intensities together.
# This makes the matrix less 'noisy' and more meaningful.
glcm_features = glcm.RadiomicsGLCM(image, mask, binWidth=25)

# 3. Calculate the Texture Features
# This does the 'Neighbor Counting' we discussed!
texture_results = glcm_features.execute()

# 4. Convert the results into a clean Pandas DataFrame
# We use orient='index' so the feature names are rows (easy to scroll)
df_texture = pd.DataFrame.from_dict(texture_results, orient='index', columns=['Value'])

# 5. Display the most famous texture markers
print("--- DAY 3: KEY TEXTURE FEATURES ---")
# Use the simple names returned by the glcm module
print(df_texture.loc[['Autocorrelation', 'Contrast', 'JointEntropy', 'Idm']])
#idm is homogeneity
# Optional: Show the top 15 texture results
# df_texture.head(15)
```
