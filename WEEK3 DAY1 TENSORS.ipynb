```python
import torch
import numpy as np

print(f"PyTorch version: {torch.__version__}")
print(f"GPU available: {torch.cuda.is_available()}")
```

```python
# Creating your first tensor
# Imagine these are HU values from a tiny CT patch

ct_patch = torch.tensor([[100, 200, 150],
                         [300, 250, 180]])

print("Tensor values:")
print(ct_patch)
print(f"\nShape: {ct_patch.shape}")
print(f"Dimensions: {ct_patch.ndim}")
print(f"Data type: {ct_patch.dtype}")
```

```python
# Shape is the most important property
# You will use .shape constantly

# A real CT slice
ct_slice = torch.zeros(512, 512)
print(f"CT slice shape: {ct_slice.shape}")

# A full CT scan (300 slices)
ct_volume = torch.zeros(300, 512, 512)
print(f"CT volume shape: {ct_volume.shape}")

# Ready for neural network
# [batch, channel, height, width]
ct_network = torch.zeros(1, 1, 512, 512)
print(f"Network ready shape: {ct_network.shape}")

# Batch of 8 CT slices
ct_batch = torch.zeros(8, 1, 512, 512)
print(f"Batch shape: {ct_batch.shape}")
```

```python
# Data types — critical for neural networks

# Integer tensor (CT HU values as loaded)
hu_values = torch.tensor([100, 200, -500, 0, 400])
print(f"Integer tensor: {hu_values}")
print(f"dtype: {hu_values.dtype}")

# Convert to float32 for neural network
hu_float = hu_values.float()
print(f"\nFloat tensor: {hu_float}")
print(f"dtype: {hu_float.dtype}")

# Segmentation mask — stays as integer
mask = torch.tensor([0, 1, 0, 0, 1])
print(f"\nMask tensor: {mask}")
print(f"dtype: {mask.dtype}")
```

```python
# Moving tensors to GPU
# This is PyTorch superpower 1

# Check device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Create tensor on CPU
ct_tensor = torch.tensor([100.0, 200.0, 300.0])
print(f"\nTensor device: {ct_tensor.device}")

# Move to GPU
ct_tensor = ct_tensor.to(device)
print(f"After .to(device): {ct_tensor.device}")

# Create tensor directly on GPU
gpu_tensor = torch.zeros(512, 512).to(device)
print(f"GPU tensor shape: {gpu_tensor.shape}")
print(f"GPU tensor device: {gpu_tensor.device}")
```

```python
# Converting between NumPy and PyTorch
# You will do this constantly
# CT data loads as NumPy → convert to tensor

import numpy as np

# Simulate CT slice loaded with pydicom
ct_numpy = np.random.randint(-1000, 400,
                              size=(512, 512),
                              dtype=np.int16)
print(f"NumPy array shape: {ct_numpy.shape}")
print(f"NumPy dtype: {ct_numpy.dtype}")
print(f"HU range: {ct_numpy.min()} to {ct_numpy.max()}")

# Convert to PyTorch tensor
ct_tensor = torch.from_numpy(ct_numpy.astype(np.float32))
print(f"\nPyTorch tensor shape: {ct_tensor.shape}")
print(f"PyTorch dtype: {ct_tensor.dtype}")

# Move to GPU
ct_tensor = ct_tensor.to(device)
print(f"Device: {ct_tensor.device}")

# Add channel and batch dimensions
# [512, 512] → [1, 1, 512, 512]
ct_tensor = ct_tensor.unsqueeze(0)  # add channel
ct_tensor = ct_tensor.unsqueeze(0)  # add batch
print(f"\nNetwork ready shape: {ct_tensor.shape}")
print("✓ CT slice ready for neural network")
```
