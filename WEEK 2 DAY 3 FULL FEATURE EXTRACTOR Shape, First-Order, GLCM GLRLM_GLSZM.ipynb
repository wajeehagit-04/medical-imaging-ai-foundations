```python
# 1. Install dependencies first
!pip install SimpleITK numpy pandas

# 2. Install the latest development version of pyradiomics from GitHub
# This version contains the fixes for Python 3.12
!pip install git+https://github.com/AIM-Harvard/pyradiomics.git@master
import SimpleITK as sitk
import pandas as pd
from radiomics import getTestCase, glcm,featureextractor


# 1. Load the sample brain data
imagePath, maskPath = getTestCase('brain1')

# 2. Initialize the "Full" Feature Extractor
# We include settings for Normalization (Day 3) and Resampling (Day 1)
settings = {
    'binWidth': 25,
    'resampledPixelSpacing': [1, 1, 1],  # Standardizes voxel size to 1mm cubes
    'interpolator': sitk.sitkBSpline,
    'normalize': True,                   # Normalizes intensity (Scanner-Agnostic)
    'normalizeScale': 100,
    'force2D': False                     # We want full 3D analysis
}

extractor = featureextractor.RadiomicsFeatureExtractor(**settings)

# 3. Enable the specific Feature Classes (Shape + FirstOrder + glcm + glrlm + glszm)
extractor.enableFeatureClassByName('shape')
extractor.enableFeatureClassByName('firstorder')
extractor.enableFeatureClassByName('glcm')   # Day 3
extractor.enableFeatureClassByName('glrlm')  # Day 4: Runs (Lines)
extractor.enableFeatureClassByName('glszm')  # Day 4: Zones (Clumps)

# 4. Run the extraction
print("Running complete extraction (Shape, First-Order, GLCM, GLRLM, GLSZM)...")
results = extractor.execute(imagePath, maskPath)

# 5. Convert results to a Clean DataFrame
df_results = pd.DataFrame.from_dict(results, orient='index', columns=['Value'])

# 6. Exact Day 4 Feature Selection
# We use the exact names confirmed by your previous run to avoid KeyErrors
day3_exact_features = [
    'original_glrlm_LongRunEmphasis',          # Measure of structural solidity
    'original_glrlm_RunEntropy',               # Measure of chaos in lines
    'original_glszm_SmallAreaEmphasis',        # Measure of speckle/fragmentation
    'original_glszm_SmallAreaHighGrayLevelEmphasis' # Hotspot detector (High Grade indicator)
]

print("\n--- DAY 3: TARGETED FEATURE ANALYSIS ---")
# Safely print the features that exist in your current dataset
print(df_results.loc[day3_exact_features])
```
